```python
import re
```

```python
# 1. Extract all email addresses from the given text.
text = """Contact us at support@example.com or sales@company.org for assistance.
For personal inquiries, email john.doe123@university.edu."""

pattern = r'[\w.-]+@[\w-]+\.[\w-]*[a-zA-Z]{2,}'
matches = re.findall(pattern, text)

for m in matches:
  print(m)
```

```python
# 2. Check if the phone numbers in the text follow the format: XXX-XXX-XXXX.
phone = "Valid: 123-456-7890, 987-654-3210\nInvalid: 12-345-67890, 1234567890, 123-45-6789"

pattern = r'\b\d{3}-\d{3}-\d{4}\b'
matches = re.findall(pattern, phone)

for m in matches:
  print(m)
```

```python
# 3. Extract all dates in the format DD/MM/YYYY or DD-MM-YYYY from the text.
dates = "Important dates: 25/12/2023, 01-01-2024, 31/05/2023, and 15-10-2024."

pattern = r'\b\d{2}[/-]\d{2}[/-]\d{4}\b'
matches = re.findall(pattern, dates)

for m in matches:
  print(m)
```

```python
# 4.  Find all instances of repeated words in the text (e.g., "the the").
text = "The the quick brown fox jumps over the the lazy dog."

pattern = r'\b((\w+)\s+\2)\b'
matches = re.findall(pattern, text, re.IGNORECASE)

for m in matches:
    print(m[0])
```

```python
# 5. Extract all hashtags (words starting with # ) from the text.
text = "Check out our new products: #Sale2024, #NewArrival, and #Discounts!"

pattern = r'#\w+'
matches = re.findall(pattern, text)

for m in matches:
  print(m)
```

```python
# 6.  Check if the passwords meet the following criteria:
#      At least 8 characters long
#      Contains at least one uppercase letter
#      Contains at least one lowercase letter
#      Contains at least one digit
text = "Valid: Password123, Secure456\nInvalid: weak, password, Password"

password = re.findall(r'\w+', text)

print("Valid passwords:")
for p in password:
  if len(p) >= 8 and re.search(r'[A-Z]', p) and re.search(r'[a-z]', p) and re.search(r'\d', p):
    print(p)
```

```python
# 7. Extract all URLs from the given text.
text = "Visit our website at https://www.example.com or check out http://blog.example.org for updates."

pattern = r'https?://\S+'
matches = re.findall(pattern, text)

for m in matches:
  print(m)
```

```python
# 8.  Replace multiple spaces in the text with a single space.
text = "This   text    has   multiple    spaces."

pattern = r'\s+'
matches = re.sub(pattern, " ", text)
print(matches)
```

```python
# 9. Extract all text within double quotes ("..." ).
text = 'He said, "Hello, world!" and she replied, "Hi there!"'

pattern = r'"(.*?)"'
matches = re.findall(pattern, text)

for m in matches:
  print(m)
```

```python
# 10. Check if the IP addresses in the text are valid (format: XXX.XXX.XXX.XXX ).
text = "Valid: 192.168.1.1, 10.0.0.255\nInvalid: 256.1.2.3, 192.168.01.1, 192.168.1"

n = r'(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)'
pattern = r'\b' + n + r'\.' + n + r'\.' + n + r'\.' + n + r'\b'
matches = re.findall(pattern, text)

for m in matches:
  print(m)
```
