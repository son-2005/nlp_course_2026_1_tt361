```python
import pandas as pd
df = pd.read_csv("Coachella-2015-2-DFE.csv", encoding = 'latin1')
print(df.head())
```

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
from nltk.tokenize import RegexpTokenizer
from nltk.corpus import stopwords
```

```python
# 1. Find all the hashtags in the tweets, such as #coachella, #beychella, etc. Extract them and save them in a separate column named "hashtags".
df["hashtags"] = df["text"].str.findall(r"#\w+")
print(df["hashtags"])
```

```python
# 2. Find all the emails in the tweets, such as example@gmail.com, user@yahoo.com, etc. Extract them and save them in a separate column named "emails".
df["emails"] = df["text"].str.findall(r"[\w.-]+@[\w.-]+\.\w+")
print(df["emails"])
```

```python
# 3. remove_usernames: This function should remove any usernames that start with @, such as @coachella, @beyonce, etc.
import re
def remove_usernames(text):
  return re.sub(r"@\w+", "", text)

df["clean_text"] = df["text"].apply(remove_usernames)

print(df["clean_text"])
```

```python
# 4. remove_links: This function should remove any links that start with http or https, such as https://www.coachella.com/, http://bit.ly/2GdOZLz, etc.
def remove_links(text):
  return re.sub(r"https?://\S+", "", text)

df["clean_text"] = df["clean_text"].apply(remove_links)
print(df["clean_text"])
```

```python
# 5. remove_non_ascii_symbols: This function should remove any symbols that are not part of the ASCII character set, such as emojis, accented letters, etc.
def remove_non_ascii_symbols(text):
  return re.sub(r"[^\x00-\x7F]", "", text)

df["clean_text"] = df["clean_text"].apply(remove_non_ascii_symbols)
print(df["clean_text"])
```

```python
# 6. to_lower: This function should convert all the letters in the tweet text to lower case, such as COACHELLA to coachella, Beyoncé to beyoncé, etc.
def to_lower(text):
  return text.lower()

df["clean_text"] = df["clean_text"].apply(to_lower)
print(df["clean_text"])
```

```python
# 7. remove_stop_words: This function should remove any common words that do not add much meaning to the tweet text, such as the, a, an, and, etc.
stop_words = stopwords.words("english")

def remove_stop_words(text):
  words = text.split()
  filtered_words = []
  for word in words:
    if word not in stop_words:
      filtered_words.append(word)

  return " ".join(filtered_words)

df["clean_text"] = df["clean_text"].apply(remove_stop_words)
print(df["clean_text"])
```

```python
# 8. remove_digits: This function should remove any numbers in the tweet text, such as 2018, 10, 100, etc.
def remove_digits(text):
  return re.sub(r"[0-9]+", "", text)

df["clean_text"] = df["clean_text"].apply(remove_digits)
print(df["clean_text"])
```

```python
# 9. remove_special_characters: This function should remove any punctuation marks or other special characters in the tweet text, such as . , ! ? # $ % & * ( ) + = - _ [ ] { } ; : ' " / \ | < > ` ~, etc.
def remove_special_characters(text):
  return re.sub(r"([^\w\s]|_)+", "", text)

df["clean_text"] = df["clean_text"].apply(remove_special_characters)
print(df["clean_text"])
```
